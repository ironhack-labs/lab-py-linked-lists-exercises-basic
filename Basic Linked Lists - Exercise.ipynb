### Exercise 1: Create a Linked List



Write a Python program to create a linked list and display its elements.

---


```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def display(self):
        """Print all elements as 1 -> 2 -> 3 -> None."""
        current, parts = self.head, []
        while current:
            parts.append(str(current.data))
            current = current.next
        print(" -> ".join(parts + ["None"]))

# Build a list 1 -> 2 -> 3 by hand
ll = LinkedList()
ll.head = Node(1)
ll.head.next = Node(2)
ll.head.next.next = Node(3)
ll.display()
```


### Exercise 2: Append a Node

Write a Python program to append a new node at the end of a linked list and display the updated list.

---


```python
def append(self, data):
    """Add a node at the end (O(n) because we walk to the tail)."""
    new_node = Node(data)
    if not self.head:
        self.head = new_node
        return
    current = self.head
    while current.next:
        current = current.next
    current.next = new_node

LinkedList.append = append  # attach the method to the class defined above
ll.append(4)
ll.display()
```


### Exercise 3: Prepend a Node

Write a Python program to prepend a new node at the beginning of a linked list and display the updated list.

---


```python
def prepend(self, data):
    """Add a node at the beginning (O(1))."""
    new_node = Node(data)
    new_node.next = self.head
    self.head = new_node

LinkedList.prepend = prepend
ll.prepend(0)
ll.display()
```


### Exercise 4: Insert a Node After a Given Node

Write a Python program that inserts a new node after a specified node in the linked list and displays the updated list.

---


```python
def insert_after(self, prev_data, data):
    """Insert a new node after the first node that holds prev_data."""
    current = self.head
    while current and current.data != prev_data:
        current = current.next
    if current is None:
        print(f"Node {prev_data} not found")
        return
    new_node = Node(data)
    new_node.next = current.next
    current.next = new_node

LinkedList.insert_after = insert_after
ll.insert_after(2, 2.5)
ll.display()
```


### Exercise 5: Delete a Node

Write a Python program to delete a specified node from the linked list and display the updated list.

---


```python
def delete(self, data):
    """Remove the first node holding data."""
    current = self.head
    if current and current.data == data:  # deleting the head
        self.head = current.next
        return
    prev = None
    while current and current.data != data:
        prev, current = current, current.next
    if current is None:
        print(f"Node {data} not found")
        return
    prev.next = current.next

LinkedList.delete = delete
ll.delete(2.5)
ll.display()
```

### Bonus Exercise 1: Reverse a Linked List

Write a Python program that reverses the linked list and displays the reversed list.

---


```python
def reverse(self):
    """Reverse the list in place by flipping every next pointer."""
    prev, current = None, self.head
    while current:
        current.next, prev, current = prev, current, current.next
    self.head = prev

LinkedList.reverse = reverse
ll.reverse()
ll.display()
```


### Bonus Exercise 2: Search for an Element

Write a Python program that searches for an element in the linked list and returns its position if found.

---


```python
def search(self, data):
    """Return the 0-based position of data or -1 if it is missing."""
    current, position = self.head, 0
    while current:
        if current.data == data:
            return position
        current, position = current.next, position + 1
    return -1

LinkedList.search = search
print("Position of 3:", ll.search(3))
print("Position of 99:", ll.search(99))
```
