### Exercise 1: Create a Linked List



Write a Python program to create a linked list and display its elements.

---


```python
# Define a Node class
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# Define a LinkedList class
class LinkedList:
    def __init__(self):
        self.head = None

    # Method to add a node at the end
    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = new_node

    # Method to display the linked list
    def display(self):
        curr = self.head
        while curr:
            print(curr.data, end=" -> ")
            curr = curr.next
        print("None")

# Example usage
ll = LinkedList()
elements = [88, -15, -22, 17, 60, 2026]
for elem in elements:
    ll.append(elem)

print("Linked List:")
ll.display()
```


### Exercise 4: Insert a Node After a Given Node

Write a Python program that inserts a new node after a specified node in the linked list and displays the updated list.

---


```python
# Define a Node class
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# Define a LinkedList class
class LinkedList:
    def __init__(self):
        self.head = None

    # Method to add a node at the end
    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = new_node

    # Method to insert after a specific value
    def insert_after(self, target_value, new_data):
        curr = self.head
        while curr:
            if curr.data == target_value:
                new_node = Node(new_data)
                new_node.next = curr.next
                curr.next = new_node
                return
            curr = curr.next
        print(f"Value {target_value} not found in the list.")

    # Method to display the linked list
    def display(self):
        curr = self.head
        while curr:
            print(curr.data, end=" -> ")
            curr = curr.next
        print("None")

# Create and populate the linked list
ll = LinkedList()
elements = [88, -15, -22, 17, 60, 2026]
for elem in elements:
    ll.append(elem)

# Insert a new node after a specified node
ll.insert_after(17, 555)

# Display the updated list
print("Updated Linked List:")
ll.display()
```


### Exercise 5: Delete a Node

Write a Python program to delete a specified node from the linked list and display the updated list.

---


```python
# Define a Node class
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# Define a LinkedList class
class LinkedList:
    def __init__(self):
        self.head = None

    # Method to add a node at the end
    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = new_node

    # Method to delete a node by value
    def delete_node(self, target_value):
        curr = self.head
        prev = None

        while curr:
            if curr.data == target_value:
                if prev:
                    prev.next = curr.next
                else:
                    self.head = curr.next  # deleting the head node
                return
            prev = curr
            curr = curr.next

        print(f"Value {target_value} not found in the list.")

    # Method to display the linked list
    def display(self):
        curr = self.head
        while curr:
            print(curr.data, end=" -> ")
            curr = curr.next
        print("None")

# Create and populate the linked list
ll = LinkedList()
elements = [88, -15, -22, 17, 60, 2026]
for elem in elements:
    ll.append(elem)

# Delete a specified node
ll.delete_node(-22)

# Display the updated list
print("Updated Linked List:")
ll.display()
```

### Bonus Exercise 1: Reverse a Linked List

Write a Python program that reverses the linked list and displays the reversed list.

---


```python
# Define a Node class
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# Define a LinkedList class
class LinkedList:
    def __init__(self):
        self.head = None

    # Method to add a node at the end
    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = new_node

    # Method to reverse the linked list
    def reverse(self):
        prev = None
        curr = self.head
        while curr:
            next_node = curr.next
            curr.next = prev
            prev = curr
            curr = next_node
        self.head = prev

    # Method to display the linked list
    def display(self):
        curr = self.head
        while curr:
            print(curr.data, end=" -> ")
            curr = curr.next
        print("None")

# Create and populate the linked list
ll = LinkedList()
elements = [88, -15, -22, 17, 60, 2026]
for elem in elements:
    ll.append(elem)

# Reverse the linked list
ll.reverse()

# Display the reversed list
print("Reversed Linked List:")
ll.display()
```


### Bonus Exercise 2: Search for an Element

Write a Python program that searches for an element in the linked list and returns its position if found.

---


```python
# Node class is defined
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# Define a LinkedList class
class LinkedList:
    def __init__(self):
        self.head = None

    # Method to add a node at the end
    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = new_node

    # Method to search for a value and return its position
    def search(self, target):
        curr = self.head
        index = 0
        while curr:
            if curr.data == target:
                return index
            curr = curr.next
            index += 1
        return -1  # Not found

    # Method to display the linked list
    def display(self):
        curr = self.head
        while curr:
            print(curr.data, end=" -> ")
            curr = curr.next
        print("None")

# Create and populate the linked list
ll = LinkedList()
elements = [88, -15, -22, 17, 60, 2026]
for elem in elements:
    ll.append(elem)

# Display the list
print("Linked List:")
ll.display()

# Search for a value
search_value = 17
position = ll.search(search_value)

# Output the result
if position != -1:
    print(f"Value {search_value} found at position {position}.")
else:
    print(f"Value {search_value} not found in the list.")
```
